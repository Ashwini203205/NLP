```python
import nltk
nltk.download('punkt_tab')

# NLTK needs a specific data package to split text into sentences. The required package (punkt_tab)
```

```python
corpus = """Hello Welcome. I am Ashwini ! please follow me. call me as ash's. """
```

```python
corpus
```

```python
print(corpus)
```

# Tokenization
#sent_tokenize


sentence --> paragraph

```python
from nltk.tokenize import sent_tokenize
```

```python
sent_tokenize(corpus)
```

# Document : type,print

```python
documents = sent_tokenize(corpus)
```

```python
type(documents)
```

```python
for sentence in documents:
  print(sentence)
```

# word_tokenize

paragraph -- >  words


sentence --> words

```python
from nltk.tokenize import word_tokenize
```

```python
word_tokenize(corpus)
```

```python
for sentence in documents:
  print(word_tokenize(sentence))
```

# wordpunct

( ' ) also consider .

```python
from nltk.tokenize import wordpunct_tokenize
```

```python
wordpunct_tokenize(corpus)
```

# TreebankWordTokenizer
not consider full stop in corpus as seprate.

only last one full stop consider.

```python
from nltk.tokenize import TreebankWordTokenizer
```

```python
tokenizer = TreebankWordTokenizer()
```

```python
tokenizer.tokenize(corpus)
```
