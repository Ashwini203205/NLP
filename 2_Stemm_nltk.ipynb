# Stremming
Stemming is chopping off the ends of words to find their base root (e.g., "talking" becomes "talk").

 It uses simple rules to drop suffixes like "-ing", "-ed", or "-s".
Here are the three main types in short:
## 1. Porter Stemmer

* Style: The classic, standard option.
* Traits: Moderately aggressive and uses simple rules.
* Result: Good for general English, but sometimes leaves words misspelled.

## 2. Lancaster Stemmer

* Style: Very aggressive.
* Traits: Choops words down heavily to the shortest possible core.
* Result: Super fast, but often creates confusing, unreadable words.

## 3. Snowball Stemmer

* Style: The modern improvement.
* Traits: Better, faster version of Porter that supports multiple languages.
* Result: The preferred choice for most modern project setups.

## 4. Regexp Stemmer

* Uses a single regular expression supplied by the user.
* Removes any prefix or suffix matching that specific pattern.
* Entirely dependent on the custom rules you write.Flexible for specific text domains or custom formats.

# 2. Regexp Stemmer Class

NLTK has RegexpStemmer class with the help of which we can easily implement Regular Expression Stemmer algorithms. It basically takes a single regular expression and removes any prefix or suffix that matches the expression. Let us see an example

```python
from nltk.stem import RegexpStemmer
```

```python
reg_stemmer=RegexpStemmer('ing$|s$|e$|able$', min=4)
```

```python
reg_stemmer.stem('eating')
```

```python
reg_stemmer.stem('ingeating')
```

```python
reg_stemmer=RegexpStemmer('ing|s$|e$|able$', min=4)     # $ in  ing  remove
reg_stemmer.stem('ingeating')
```
